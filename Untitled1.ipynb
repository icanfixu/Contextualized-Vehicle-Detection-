```python
import matplotlib.pyplot as plt

class_map = metrics.box.maps

plt.figure(figsize=(10,5))
plt.bar(range(len(class_map)), class_map)
plt.xlabel("Class Index")
plt.ylabel("mAP")
plt.title("Per-Class mAP (Validation Dataset)")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("runs/detect/train/results.csv")
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["metrics/mAP50(B)"], label="Validation mAP50")
plt.plot(df["metrics/mAP50-95(B)"], label="Validation mAP50-95")

plt.xlabel("Epoch")
plt.ylabel("mAP")
plt.title("Validation mAP vs Epoch")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["metrics/precision(B)"], label="Precision")
plt.plot(df["metrics/recall(B)"], label="Recall")

plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title("Validation Precision & Recall vs Epoch")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["val/box_loss"], label="Validation Box Loss")
plt.plot(df["val/cls_loss"], label="Validation Class Loss")
plt.plot(df["val/dfl_loss"], label="Validation DFL Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Validation Loss vs Epoch")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("runs/detect/train/results.csv")
```

```python
plt.figure(figsize=(8,5))

plt.plot(df["metrics/mAP50(B)"], label="Validation mAP50")
plt.plot(df["metrics/mAP50-95(B)"], label="Validation mAP50-95")

plt.xlabel("Epoch")
plt.ylabel("mAP")
plt.title("Training Progress (Validation mAP)")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(df["train/box_loss"], label="Train Box Loss")
plt.plot(df["val/box_loss"], label="Validation Box Loss")

plt.plot(df["train/cls_loss"], label="Train Class Loss")
plt.plot(df["val/cls_loss"], label="Validation Class Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(df["metrics/precision(B)"], label="Precision")
plt.plot(df["metrics/recall(B)"], label="Recall")

plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title("Precision vs Recall During Training")
plt.legend()
plt.grid(True)

plt.show()
```
